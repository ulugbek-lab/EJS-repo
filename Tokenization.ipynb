```python
text = 'hello this is my page and my google colab' # around 9 tokens
print(text)
```

```python
tokens = text.split()
print(tokens)
```

```python
vocab = {}
current_id = 0

for token in tokens:
  if token not in vocab:
    vocab[token] = current_id
    current_id += 1

print(vocab)
```

```python
encoded = [vocab[token] for token in tokens]
print(encoded)
```

```python
reverse_vocab = {id: token for token, id in vocab.items()}
decoded = " ".join([reverse_vocab[id] for id in encoded])
print(decoded)
```

```python
# Embedding vector -> dense numerical array || Vector Database
# '1' => [0.41, 0.22, -0.77, 0.56.......]
# '0' => [0.78, 0.21, -0.67, 0.12.......]
import torch
import torch.nn as nn

embedding_dim = 8
vocab_size = len(vocab)

embedding_layer = nn.Embedding(vocab_size, embedding_dim)

imput_ids = torch.tensor([encoded])
embeddings = embedding_layer(imput_ids)

print(embeddings)
print(embeddings.shape)



```
